customersテーブルに対して，以下のようなSQL文を作成せよ

(1) customer_ageカラムの平均を求めるSQL文

```python
%%sql
SELECT avg(customer_age) FROM customers;
```

(2) customer_ageカラムの最大値を求めるSQL文

```python
%%sql
SELECT max(customer_age) FROM customers;
```

(3) customer_ageカラムの値の重複を除いた行の数を求めるSQL文

(4) customer_genderカラムの値が'F'であるような行について，customer_ageカラムの平均を求めるSQL文

# 例題：GROUP BY句の利用

customersテーブルに対して，以下のようなSQL文を作成せよ

(5) customer_genderカラムの値と，customer_genderカラムの値（性別）毎のcustomer_ageカラム（年齢）の平均の値の組をすべてを求めるSQL文

```python
%%sql
SELECT customer_gender, avg(customer_age) FROM customers
GROUP BY customer_gender;
```

(6) customer_ageカラムの値と，customer_ageカラムの値毎の行数との組（つまり，年齢毎の顧客数）を，customer_ageカラムの値毎の行数が2000より大きい組だけを求めるSQL文

```python
%%sql
SELECT customer_age, count(*) FROM customers
GROUP BY customer_age HAVING count(*) > 2000;
```
