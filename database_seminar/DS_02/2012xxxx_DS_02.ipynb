## SQliteを用いたデータベース作成

このノートブックでは、Python環境（Jupyter/Colab）で `jupysql` を使って、SQLを実行する方法を学びます。

- SQLite（オンメモリ）を使って、簡単なデータベースを作成します。
- 学生（`学生`）と講義（`講義`）の2つのテーブルを用意し、日本語のサンプルデータを登録します。
- 基本的な検索（`SELECT`）や結合（`JOIN`）を体験してみましょう。


```
# jupysqlのインストール
!pip install jupysql --quiet
%load_ext sql
%sql sqlite://
```

```
%sql sqlite://
```

```
%%sql

CREATE TABLE 学生 (
    学籍番号 INTEGER PRIMARY KEY,
    氏名 TEXT,
    学年 INTEGER
);
```

```
%%sql

CREATE TABLE 講義 (
    講義ID INTEGER PRIMARY KEY,
    講義名 TEXT,
    学籍番号 INTEGER,
    FOREIGN KEY(学籍番号) REFERENCES 学生(学籍番号)
);
```

```
%%sql

INSERT INTO 学生 (学籍番号, 氏名, 学年) VALUES
(1, '佐藤 花子', 1),
(2, '鈴木 太郎', 2),
(3, '高橋 一郎', 1),
(4, '田中 美咲', 3),
(5, '伊藤 健', 2);
```

```
%%sql

INSERT INTO 講義 (講義ID, 講義名, 学籍番号) VALUES
(101, 'データサイエンス入門', 1),
(102, 'プログラミング基礎', 2),
(103, '統計学', 1),
(104, '情報倫理', 3),
(105, '体育', 5);
```

```
%%sql
SELECT * FROM 学生;
```

```
%%sql
SELECT 氏名, 学年 FROM 学生 WHERE 学年 >= 2;
```

```
%%sql
SELECT 学生.氏名, 講義.講義名
FROM 学生
JOIN 講義 ON 学生.学籍番号 = 講義.学籍番号;
```

## テーブル情報（スキーマ）

### テーブル：教員

| カラム | データ型 | 制約・備考 |
|---|---|---|
| 教員番号 | INTEGER | 主キー |
| 教員名 | TEXT | |
| 職位 | TEXT | 教授・准教授などの値をとる |
| 学部名 | TEXT | 今回は全行同じ値になる |
| 教員説明文 | TEXT | 空欄または簡潔なメモで可 |

### テーブル：科目

| カラム | データ型 | 制約・備考 |
|---|---|---|
| 科目番号 | INTEGER | 主キー |
| 科目名 | TEXT | |
| 担当教員番号 | INTEGER | 教員テーブルの主キーを参照する**外部キー** |

* 一つの科目を複数の教員が担当することがあるが，単純化のため一つの科目は一人の教員が担当することとする（担当者の誰か一人を設定する）
* 科目の単位はホームページ上の表記を一つの単位とする（例：「アニメ制作実務Ⅰ・Ⅱ」は1科目として扱う）

## データの要件

* **教員：任意の5人以上**（情報学部の教員から選ぶ）
* **科目：5個以上**（外部キーで教員テーブルの教員番号と正しく対応させること）
