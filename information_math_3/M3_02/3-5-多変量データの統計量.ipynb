## 5章　多変量データの統計量

### 実装：分析の準備

```python
# 数値計算に使うライブラリ
import numpy as np
import pandas as pd
```

### 実装：分析対象となるデータの用意

```python
# データの読み込み
cov_data = pd.read_csv('3-5-1-cov.csv')
print(cov_data)
```

### 実装：共分散

```python
# データの取り出し
x = cov_data['x']
y = cov_data['y']

# サンプルサイズ
n = len(cov_data)

# 標本平均
x_bar = np.mean(x)
y_bar = np.mean(y)
```

```python
# 共分散
cov = sum((x - x_bar) * (y - y_bar)) / n
round(cov, 3)
```

### 実装：分散共分散行列

```python
# 分散の計算
s2_x = np.var(x, ddof=0)
s2_y = np.var(y, ddof=0)

print('xの標本分散：', round(s2_x, 3))
print('yの標本分散：', round(s2_y, 3))
```

```python
# 共分散
np.cov(x, y, ddof=0)
```

### 実装：ピアソンの積率相関係数

```python
# 相関係数
rho = cov / np.sqrt(s2_x * s2_y)
round(rho, 3)
```

```python
# 相関行列
np.corrcoef(x, y)
```

#### クロス集計表の作成

```python
shoes = pd.read_csv('3-5-3-cross2.csv')
print(shoes)
```

```python
cross_2 = pd.pivot_table(
    data=shoes,
    values='sales',
    aggfunc='sum',
    index='store',
    columns='color'
)
print(cross_2)
```
