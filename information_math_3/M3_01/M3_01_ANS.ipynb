## 演習
上のセルで取得したデータファイル（8-1-1-beer.csv）で，temperatureが30以上であるbeerの値をすべて表示せよ．教科書p.55〜と同様にquery関数を用いること．

```python
import pandas as pd

df = pd.read_csv("8-1-1-beer.csv")
df.head()
```

```python
# temperetureが30以上であるbeerの値をquery関数を用いて求めよ
df.query("temperature >= 30")["beer"]
```

```python
# temperetureが30以上であるbeerの値を求めよ（参考：query関数を使わない方法）
df[df["temperature"] >= 30]["beer"]
```
