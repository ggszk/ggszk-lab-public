## 例題：見つけたテントウムシは新種？

* ある日，A君がキャンパス内で，見たことのないテントウムシを1匹発見した．
* しかし，B君に見せたところ，「背の模様が同じなので◯◯テントウという種だ」と言われた．
* 果たして，A君が見つけたテントウムシは，◯◯テントウか？新しい種なのか？
* 得られている情報：
  * A君が採集したテントウムシの（平均）体長は，11mm
  * ◯◯テントウの体長は，5mm，母標準偏差は，2（mm）

### 限界値を求める
* 標準化しないやり方でやってみよう
* 資料の「検定のステップ3 確率の計算：限界値を求める」

```python
# 母集団に関する情報
mu = 5         # ◯◯テントウの体長（母平均）
sigma = 2      # ◯◯テントウの母標準偏差（1匹のばらつき）

# 標本に関する情報
x_bar = 11 # 観察された標本平均（1匹の体長）
n = 1 # 標本サイズ

# 標準誤差（標本平均のばらつき）
se = sigma / np.sqrt(n)

# 標準正規分布の2.5%点と97.5%点
z_025 = norm.ppf(0.025)
z_975 = norm.ppf(0.975)

# 上下の限界値の計算
lower_mu = mu + z_025 * se
upper_mu = mu + z_975 * se

# 小数第3位で表示
print(f"z_025 = {round(z_025, 3)}, z_975 = {round(z_975, 3)}")
print("下側限界値：", round(lower_mu, 3))
print("上側限界値：", round(upper_mu, 3))
```

### 結論（判定方法1：限界値より外か内か）
* 上記の計算から，標本平均11は上側限界値よりも大きく，棄却域にあるため，平均体長には有意な差がある．

### p値を求める

* p値を求めてみよう
* p値は，実験結果（この場合は11）よりも極端な値が観測される確率
* まずは，標準正規分布ではなく，〇〇テントウの標本分布で考えてみよう
* 両側にあることに注意する
* p値の値によって，結論の文章を出力することができる

```python
# p値を求める
p_upper = 1 - norm(mu, se).cdf(x_bar) # 正規分布の累積分布関数を使用．右側半分の確率を求める
#p_upper = 1 - norm(loc=mu, scale=se).cdf(x_bar) # パラメータを明示的に書く方法
p_value = 2 * p_upper

print(f"p値: {p_value:.4f}")

# 判定（有意水準5%）
alpha = 0.05
if p_value < alpha:
    print("p値が有意水準より小さいため、帰無仮説は棄却される。")
    print("この結果は、統計的に有意である。")
else:
    print("p値が有意水準以上のため、帰無仮説は棄却できない。")
    print("この結果は、統計的に有意ではない。")
```

### 結論（判定方法2：p値を使用）
* 上記の計算から，p値が有意水準5%より小さいために，平均体長には有意な差がある

### 標準正規分布を使用する場合
* 標準化した値（Z値とも呼ばれる）を求めてみよう
* 限界値とp値を求めてみよう
* 結論は上記と同じである．p値が同じであることを確認しよう

```python
# Z値の計算
z = (x_bar - mu) / se

# p値の計算
p_value = 2 * (1 - norm.cdf(abs(z))) # 標準正規分布の累積分布関数を使用

print(f"Z値: {z:.3f}")
print(f"p値: {p_value:.4f}")

# 判定（有意水準5%）
alpha = 0.05
if p_value < alpha:
    print("p値が有意水準より小さいため、帰無仮説は棄却される。")
    print("この結果は、統計的に有意である。")
else:
    print("p値が有意水準以上のため、帰無仮説は棄却できない。")
    print("この結果は、統計的に有意ではない。")
```

### 参考
* 母平均の検定で，t検定を使う場合は，scipy.stats.ttest_1sampという便利な関数があり，呼び出すだけで結果を出すことができる．
* ただし，標本データの値が全て手元にある必要がある（標本平均と標本標準誤差だけではできない）．
* 上記でやったような正規分布を使う場合には，そのような便利な関数はないため，自力で計算する必要がある．
* scipy.stats.ttest_1sampの使い方は，教科書のサンプルコードに例があるので参照すること

## 演習

Aくんは，地元の販売農家100人に対してアンケート調査を実施し，
* 平均年齢：65歳，標準偏差：18（歳）

の結果を得た．ちなみに，日本の販売農家は
* 平均年齢：63歳，標準偏差：11（歳）

であることが知られている．Aくんの調査した農家は，高齢者に偏ってしまったというべきであろうか．5%の有意水準で検定せよ．100人と標本サイズが大きいため，標本分布が正規分布に従うことを前提とせよ

```python
# 母集団に関する情報
mu = 63     # 全国平均（母平均）
sigma = 11  # 全国の標準偏差（母標準偏差）

# 標本に関する情報
x_bar = 65  # 観察された標本平均
n = 100     # 標本サイズ

# 標準誤差の計算
se = sigma / np.sqrt(n)

# Z値の計算
z = (x_bar - mu) / se

# p値（両側検定）
p_value = 2 * (1 - norm.cdf(abs(z)))

print(f"Z値: {z:.3f}")
print(f"p値: {p_value:.4f}")

# 判定（有意水準5%）
alpha = 0.05
if p_value < alpha:
    print("p値が有意水準より小さいため、帰無仮説は棄却される。")
    print("この結果は、統計的に有意である。")
else:
    print("p値が有意水準以上のため、帰無仮説は棄却できない。")
    print("この結果は、統計的に有意ではない。")
```
