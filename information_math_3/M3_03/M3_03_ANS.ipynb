```python
# データの展開方法（スマートなやり方）
df3 = pd.melt(df, id_vars=['ポット番号'], value_vars=['栽培法A','栽培法B'], var_name='栽培法', value_name='収量')
```

```python
sns.boxplot(x='栽培法', y='収量',
            data=df3)
```
