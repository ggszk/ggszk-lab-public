## 例題1：アルゴリズムの性能比較

「30個のデータセットを無作為に2つのグループに分け，それぞれにアルゴリズムAまたはBを適用し，処理時間（ミリ秒）を測定した．」
* データセット1〜15：アルゴリズムA適用
* データセット16〜30：アルゴリズムB適用

結果のデータは以下のようにまとめている（次のセル（辞書のリスト形式））．

| キー名   | 説明                                       | データ型 | 単位       |
|----------|--------------------------------------------|----------|------------|
| group    | アルゴリズムのグループ（"A" または "B"）  | 文字列   | -          |
| dataset  | データセット番号（1～30）                  | 整数     | -          |
| time     | アルゴリズムの処理時間                     | 整数     | ミリ秒(ms) |

アルゴリズムAとアルゴリズムBの平均的な処理時間に有意な差があるかどうかを，有意水準5%で検定せよ．

注意：2つのグループのデータセットに対応はないので，対応のないt検定を用いる．

```python
# アルゴリズム（AとB）の処理時間
data = [
    {"group": "A", "dataset": 1, "time": 3063},
    {"group": "A", "dataset": 2, "time": 2275},
    {"group": "A", "dataset": 3, "time": 2089},
    {"group": "A", "dataset": 4, "time": 2855},
    {"group": "A", "dataset": 5, "time": 2836},
    {"group": "A", "dataset": 6, "time": 3219},
    {"group": "A", "dataset": 7, "time": 2817},
    {"group": "A", "dataset": 8, "time": 2136},
    {"group": "A", "dataset": 9, "time": 2540},
    {"group": "A", "dataset": 10, "time": 2263},
    {"group": "A", "dataset": 11, "time": 2140},
    {"group": "A", "dataset": 12, "time": 1757},
    {"group": "A", "dataset": 13, "time": 2499},
    {"group": "A", "dataset": 14, "time": 2093},
    {"group": "A", "dataset": 15, "time": 2073},
    {"group": "B", "dataset": 16, "time": 3157},
    {"group": "B", "dataset": 17, "time": 2707},
    {"group": "B", "dataset": 18, "time": 3270},
    {"group": "B", "dataset": 19, "time": 3181},
    {"group": "B", "dataset": 20, "time": 3633},
    {"group": "B", "dataset": 21, "time": 3404},
    {"group": "B", "dataset": 22, "time": 2219},
    {"group": "B", "dataset": 23, "time": 2730},
    {"group": "B", "dataset": 24, "time": 3408},
    {"group": "B", "dataset": 25, "time": 3203},
    {"group": "B", "dataset": 26, "time": 2938},
    {"group": "B", "dataset": 27, "time": 3286},
    {"group": "B", "dataset": 28, "time": 2920},
    {"group": "B", "dataset": 29, "time": 3332},
    {"group": "B", "dataset": 30, "time": 3478}
]
```

### データフレームにする


```python
df = pd.DataFrame(data)
df.head()
```

### 資料通りに計算過程も示す　→ 簡単な方法を後に示す．演習や小テストはそちらを使うと良い

### 検定統計量（t値）の計算

```python
# AのグループとBのグループを作る
time_a = df.query("group == 'A'")["time"]
time_b = df.query("group == 'B'")["time"]

# グループ毎の平均値
x_bar_a = np.mean(time_a)
x_bar_b = np.mean(time_b)

# 分散
u2_a = np.var(time_a, ddof=1)
u2_b = np.var(time_b, ddof=1)

# サンプルサイズ
n = len(time_a)
#n = len(time_b) # サンプルサイズは同じなのでどちらか一方で良い

# t値（検定統計量）
t_value = (x_bar_a - x_bar_b) / np.sqrt((u2_a/n + u2_b/n))
print(round(t_value, 3))
```

### 自由度を求める
* 2群の分散が等しい仮定（等分散の仮定）を置くとする
* その場合の自由度は，$dof = (n_1-1) + (n_2-1)$で求められる（dfをデータフレームの変数名で使ったのでdofとした）
* 分散が等しくない場合の計算は，教科書付属のノートブックを参照すること

```python
# 自由度
dof = (n - 1)+ (n - 1)
dof
```

### p値を求める→結論

```python
# p値の計算
p_value = 2 * (1 - t.cdf(abs(t_value), df=dof)) 

print(f"p値: {p_value:.7f}")

# 判定（有意水準5%）
alpha = 0.05
if p_value < alpha:
    print("p値が有意水準より小さいため、帰無仮説は棄却される。")
    print("この結果は、統計的に有意である。")
else:
    print("p値が有意水準以上のため、帰無仮説は棄却できない。")
    print("この結果は、統計的に有意ではない。")
    
```

### 簡単な方法：演習や小テストではこちらの利用を推奨
* stats.ttest_ind関数を使うと一発で求められる
* 結論は上記と同じ

```python
# 検定の実施
# equal_var=Trueは等分散を仮定
t_stat, p_value = stats.ttest_ind(time_a, time_b, equal_var=True) 
print(f"p値: {p_value:.7f}")
print(f"t値: {t_stat:.3f}")

# 判定（有意水準5%）
alpha = 0.05
if p_value < alpha:
    print("p値が有意水準より小さいため、帰無仮説は棄却される。")
    print("この結果は、統計的に有意である。")
else:
    print("p値が有意水準以上のため、帰無仮説は棄却できない。")
    print("この結果は、統計的に有意ではない。")
```

## 例題2：システム改善前後の処理性能の比較

ある計算システムに対して最適化（例えばソフトウェア更新やハードウェアチューニング）を行い，改善前後の処理効率（単位処理あたりのリソース消費の少なさなど）を5台の装置で測定した．

| キー名   | 説明                                   | データ型 | 単位       |
|----------|----------------------------------------|----------|------------|
| device   | 装置の識別番号                         | 整数     | -          |
| before   | 最適化前の処理性能スコア              | 浮動小数 | スコア単位（1が基準） |
| after    | 最適化後の処理性能スコア              | 浮動小数 | スコア単位（1が基準） |

システムの最適化処理により，処理性能が有意に変化したと言えるか．有意水準5%で検定せよ

注意：同じ装置の最適化前後なので，対応のあるt検定を使う

→ ノートブック参照

```python
# デバイスごとの処理性能スコアのデータ
data_system = [
    {"device": 1, "before": 0.8, "after": 1.1},
    {"device": 2, "before": 0.7, "after": 1.5},
    {"device": 3, "before": 1.0, "after": 1.0},
    {"device": 4, "before": 0.5, "after": 1.6},
    {"device": 5, "before": 0.6, "after": 1.3}
]
```

### 検定する



```python
df_system = pd.DataFrame(data_system)
before = df_system["before"]
after = df_system["after"]

# 対応のあるt検定（まずはどんな結果がでるのかを確かめる）
stats.ttest_rel(after, before)
```

### 結論を出す

```python
# 判定（有意水準5%）
alpha = 0.05
t_stat, p_value = stats.ttest_rel(after, before)
print(f"対応のあるt検定のp値: {p_value:.7f}")

if p_value < alpha:
    print("p値が有意水準より小さいため、帰無仮説は棄却される。")
    print("この結果は、統計的に有意である。")
else:
    print("p値が有意水準以上のため、帰無仮説は棄却できない。")
    print("この結果は、統計的に有意ではない。")
```

## 例題3：等分散の検定

例題1（アルゴリズムの性能比較）では，等分散を前提とした検定を行ったが，この2群について，有意水準5%で等分散の検定を行い，等分散を前提としてよいかを検証せよ．

注意：分散比は大きい方を分子として（F値は1より大きくなる），（右側の）片側検定を実施する．

```python
# 等分散の検定
# time_aとtime_bは例題1で定義したAとBの処理時間データ

# 不偏分散（再計算）
u2_a = np.var(time_a, ddof=1)
u2_b = np.var(time_b, ddof=1)

# 分散比（大きい方を分子）
if u2_a > u2_b:
    F = u2_a / u2_b
    dof1, dof2 = len(time_a) - 1, len(time_b) - 1
else:
    F = u2_b / u2_a
    dof1, dof2 = len(time_b) - 1, len(time_a) - 1

# p値（右側の片側検定：P(F ≥ f)）
p_value = 1 - stats.f.cdf(F, dof1, dof2)

# 結果表示
print(f"F値（分散比）: {F:.3f}")
print(f"p値（片側, P(F ≥ f)）: {p_value:.3f}")

# 判定（有意水準5%）
alpha = 0.05
if p_value < alpha:
    print("p値が有意水準より小さいため、帰無仮説（分散が等しい）は棄却される。")
    print("等分散を前提とせず、ウェルチの検定を使う。")
else:
    print("p値が有意水準以上のため、帰無仮説は棄却できない。")
    print("等分散を前提としてよい（例題1のt検定のままでよい）。")
```

### 参考：Leveneの検定
* 資料に掲載したシンプルなF分布を使う方法と異なるが，同じく分散が等しいかを確認する検定
* stats.levene関数で一発で求められる

```python
# 等分散の検定
# time_aとtime_bは例題1で定義したAとBの処理時間データ
stat, p_value = stats.levene(time_a, time_b)

# 結果の表示
print(f"Levene検定統計量: {stat:.3f}")
print(f"p値: {p_value:.3f}")

# 判定
if p_value < 0.05:
    print("有意差あり：等分散とは言えない（不等分散）")
else:
    print("有意差なし：等分散とみなしてよい")
```
