```python
# 使用するライブラリのインポート
import numpy as np
import pandas as pd
from scipy.stats import norm
from scipy.stats import t
from scipy.stats import chi2
```

## 例題：正規分布による区間推定

ある園芸農家が，母の日に向けて出荷を控えたカーネーション16本の蕾（つぼみ）の直径を調べたところ，平均で10.0mmだった．この園芸農家の栽培しているカーネーションの蕾の平均直径を信頼係数95%で推定せよ．ただし，母分散は36.0($mm^2$)であるとわかっているとする．

### 解説
* 教科書p.297（これはt分布を使う例）に近い変数名に揃えた
* 信頼限界の結果は，小数点3桁で結果を丸めている（round関数を使用）
* 教科書では，lower_mu = x_bar - t_975 * se としていた．↓の方が自然かと思うが．結果は同じなのでお好みで．

```python
# 与えられたデータ
x_bar = 10.0         # 標本平均
sigma_squared = 36.0 # 母分散
n = 16               # 標本サイズ

# 母標準偏差（母分散の平方根）
sigma = np.sqrt(sigma_squared)

# 標準誤差（standard error）
se = sigma / np.sqrt(n)

# 標準正規分布の2.5%点と97.5%点
z_025 = norm.ppf(0.025)
z_975 = norm.ppf(0.975)

# 信頼区間の計算
lower_mu = x_bar + z_025 * se
upper_mu = x_bar + z_975 * se

# 小数第3位で表示
print(f"母標準偏差 sigma = {round(sigma, 3)}")
print(f"標準誤差 se = {round(se, 3)}")
print(f"z_025 = {round(z_025, 3)}, z_975 = {round(z_975, 3)}")
print("下側信頼限界：", round(lower_mu, 3))
print("上側信頼限界：", round(upper_mu, 3))
```

## 例題：t分布による区間推定

ある日，ある酪農家が，搾乳中のホルスタイン5頭の乳量を調べたら，1頭あたりの平均乳量は22.1リットル，不偏標準偏差は7.267だった．この農家が飼養しているホルスタインの1頭あたりの乳量（/日）を，信頼係数95%で推定せよ．

### 解説
* 教科書p.296-297と同様

```python
# 与えられたデータ
n = 5            # 標本サイズ
df = n - 1       # 自由度
u = 7.267        # 不偏標準偏差（分母 n-1）
x_bar = 22.1     # 標本平均（リットル）

# 標準誤差（standard error）
se = u / np.sqrt(n)

# t分布の2.5%点と97.5%点
t_025 = t.ppf(0.025, df)
t_975 = t.ppf(0.975, df)

# 信頼区間の計算
lower_mu = x_bar + t_025 * se
upper_mu = x_bar + t_975 * se

# 結果表示（小数第3位で四捨五入）
print(f"標準誤差 se = {round(se, 3)}")
print(f"t_025 = {round(t_025, 3)}, t_975 = {round(t_975, 3)}")
print("下側信頼限界：", round(lower_mu, 3))
print("上側信頼限界：", round(upper_mu, 3))
```

## 例題：分散の区間推定

Aという種のテントウムシを5匹採集した．それらの体長は，小さい順に5mm, 8mm, 10mm, 11mm, 15mm だった．このテントウムシの体長の母分散に対する信頼区間を信頼係数95%で推定せよ

```python
# データ
x = np.array([5, 8, 10, 11, 15]) # 標本データ
n = len(x) # 標本サイズ
df = n - 1 # 自由度

# 不偏分散 u^2 の計算
u2 = np.var(x, ddof=1)

# カイ二乗分布の2.5%点と97.5%点
chi2_025 = chi2.ppf(0.025, df)
chi2_975 = chi2.ppf(0.975, df)

# 母分散の信頼区間の計算
lower_var = (df * u2) / chi2_975
upper_var = (df * u2) / chi2_025

# 小数第3位で四捨五入して出力
print(f"不偏分散 u^2 = {round(u2, 3)}")
print(f"自由度 df = {df}")
print(f"chi2_025 = {round(chi2_025, 3)}, chi2_975 = {round(chi2_975, 3)}")
print("下側信頼限界（母分散）：", round(lower_var, 3))
print("上側信頼限界（母分散）：", round(upper_var, 3))
```
