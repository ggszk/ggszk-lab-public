## 例題：不偏分散の計算

* 標準正規分布の母集団からの標本サイズ10で標本を1つ作る
* 標本分散と不偏分散を計算して，母分散と比較してみよう
  * np.varを使う．ddofの値が0→標本分散，1→不偏分散（ddofは自由度からの差分を意味する）
* 標本サイズを変化させるとどう変わるか確かめてみよう

```python
# 母集団：標準正規分布（平均0, 標準偏差1）
mu = 0
sigma = 1
n = 10  # 標本サイズ

# 標本を1つ生成（標準正規分布から n 個）
sample = norm.rvs(loc=mu, scale=sigma, size=n)

# 分散の計算
sample_var = np.var(sample, ddof=0)      # 分母 n：標本分散
sample_unbiased_var = np.var(sample, ddof=1)  # 分母 n-1：不偏分散

# 出力
print("標本:", sample)
print(f"標本分散（分母 n）: {sample_var:.4f}")
print(f"不偏分散（分母 n-1）: {sample_unbiased_var:.4f}")
print(f"母分散（理論値）: {sigma**2:.1f}")
```

## 例題：不偏分散を理解する

* 標準正規分布の母集団からの標本サイズ10，標本数100で標本を作る
* 標本分散，不偏分散を求めてヒストグラムを書いてみる
* 母集団の分散にどちらが近い？

```python
# 母集団: 標準正規分布（平均0, 分散1）
mu = 0
sigma = 1

# パラメータ設定
n = 10        # 標本サイズ
n_samples = 100  # 標本数

# 分散を保存する配列
sample_vars_n = []     # 分母n（標本分散）
sample_vars_n1 = []    # 分母n-1（不偏分散）

# 標本を複数回生成して分散を計算
for _ in range(n_samples):
    sample = np.random.normal(mu, sigma, n)
    var_n = np.var(sample, ddof=0)  # 分母n
    var_n1 = np.var(sample, ddof=1) # 分母n-1（不偏分散）
    sample_vars_n.append(var_n)
    sample_vars_n1.append(var_n1)

# ヒストグラムの描画
plt.hist(sample_vars_n, bins=15, alpha=0.6, label='標本分散（ddof=0）')
plt.hist(sample_vars_n1, bins=15, alpha=0.6, label='不偏分散（ddof=1）')

# 母分散の線（σ^2 = 1）
plt.axvline(x=1, color='k', linestyle='dashed', linewidth=1, label='母分散 (1.0)')

# ラベルと凡例
plt.xlabel('分散の値')
plt.ylabel('頻度')
plt.title(f'標本分散 vs 不偏分散（標本サイズ {n}, 試行数 {n_samples}）')
plt.legend()
plt.grid(True)
plt.show()
```
