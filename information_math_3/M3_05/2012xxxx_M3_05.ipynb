## 例題：標本分布を体験してみよう

* 標本サイズを4，標本数を30とする
* 正規分布の乱数を生成する関数scipy.stats.norm.rvsを用いて，標本（サンプル）を作ろう
* 標本から標本平均を求める
* ヒストグラムを書く

```python
# 母集団の平均と標準偏差
mu = 0
sigma = 1

# 標本サイズ
sample_size = 4
n_trials = 30  # 標本数（= 平均の数）

# 階級の範囲
bins = np.linspace(-1, 1, 11) # -1から1までの範囲を10等分（11個のビン）

# 標本平均を格納するリスト
sample_means = []  
# 標本平均の分布を作成
for i in range(n_trials) :
    sample = norm.rvs(loc=mu, scale=sigma, size=sample_size)
    sample_mean = np.mean(sample)
    sample_means.append(sample_mean)

sns.histplot(sample_means, bins=bins)
plt.title(f"Distribution of Sample Means (n = {sample_size}, {n_trials} trials)")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.show()
```
